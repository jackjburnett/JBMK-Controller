```python
# Library imports
import cadquery as cq
```

```python
# Create the mount's variables
mount_values = {
    "height": 6.0,
    "diameter": 6.0,
    "X_point_width": 4.2,
    "X_point_length": 1.4,
}
```

```python
mount_X_point = (
    cq.Workplane()
    .rect(mount_values["X_point_width"], mount_values["X_point_length"])
    .rect(mount_values["X_point_length"], mount_values["X_point_width"])
    .extrude(mount_values["height"])
)
display(mount_X_point)
```

```python
mount = (
    cq.Workplane()
    .circle(mount_values["diameter"] / 2)
    .extrude(mount_values["height"])
    .cut(mount_X_point)
)
display(mount)
```

```python
# Set cap's variables
cap_values = {
    "diameter": 12.0,
    "thickness": 2.0,
    "bevel": True,
    "wall": {"thickness": 1.0, "height": 1.0}
}
```

```python
top = cq.Workplane().circle(cap_values["diameter"] / 2)
top = top.extrude(cap_values["thickness"])
display(top)
```

```python
if cap_values["bevel"]:
    top = top.faces(">Z").edges().fillet(1)
display(top)
```

```python
cap = cq.Assembly().add(top, loc=cq.Location((0, 0, cap_values["wall"]["height"])))
display(cap)
```

```python
if cap_values["wall"]["thickness"] > 0.0 and cap_values["wall"]["height"] > 0.0:
    walls = (
        cq.Workplane()
        .circle(cap_values["diameter"] / 2)
        .circle((cap_values["diameter"] / 2) - cap_values["wall"]["thickness"])
        .extrude(cap_values["wall"]["height"] + (cap_values["thickness"] / 2))
    )
    cap.add(walls)
display(cap)
```

```python
cap.add(mount, loc=cq.Location((0, 0, cap_values["wall"]["height"]-(mount_values["height"]))))
display(cap)
```

```python
cap.save("12mm_button.step")
```
